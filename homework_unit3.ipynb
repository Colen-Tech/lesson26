```python
def calculate_bmi(weight, height, system):
    """Calculates BMI based on the chosen unit system."""
    if height <= 0:
        return None

    if system == 'metric':
        # Formula: weight (kg) / [height (m)]^2
        bmi = weight / (height ** 2)
    else:
        # Formula: 703 x weight (lbs) / [height (in)]^2
        bmi = 703 * (weight / (height ** 2))

    return round(bmi, 2)

def get_bmi_category(bmi):
    """Returns the standard adult BMI category."""
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 24.9:
        return "Normal weight"
    elif 25 <= bmi < 29.9:
        return "Overweight"
    else:
        return "Obesity"

def main():
    print("Welcome to the BMI Calculator")

    try:
        # Get demographic info
        age = int(input("Enter your age (e.g., 25): "))
        gender = input("Enter your gender (e.g., Male, Female, Other): ").strip().capitalize()

        # Choose unit system
        print("\nWhich units would you like to use?")
        print("1. Metric (kilograms & meters)")
        print("2. US / Imperial (pounds & feet/inches)")
        unit_choice = input("Enter 1 or 2: ").strip()

        if unit_choice == '1':
            system = 'metric'
            weight = float(input("\nEnter your weight in kg (e.g., 65.5): "))
            height = float(input("Enter your height in meters (e.g., 1.75): "))

        elif unit_choice == '2':
            system = 'us'
            weight = float(input("\nEnter your weight in pounds (e.g., 150): "))
            feet = float(input("Enter your height - feet (e.g., 5): "))
            inches = float(input("Enter your height - inches (e.g., 9): "))
            # Convert feet and inches to total inches for the formula
            height = (feet * 12) + inches

        else:
            print("\nInvalid choice. Please run the program again and select 1 or 2.")
            return

        # Calculate and display results
        bmi_value = calculate_bmi(weight, height, system)

        if bmi_value is not None:
            category = get_bmi_category(bmi_value)

            print("\n--- Your Results ---")
            print(f"Age: {age}")
            print(f"Gender: {gender}")
            print(f"BMI: {bmi_value}")
            print(f"Category: {category}")

            if age < 18:
                print("\n*Note: Standard BMI categories are designed for adults (18+).")
                print(" For children and teens, doctors typically use age and gender-specific growth charts.")

        else:
            print("\nError: Height must be greater than zero.")

    except ValueError:
        print("\nInvalid input. Please make sure you are entering numbers for your measurements.")

if __name__ == "__main__":
    main()
```
