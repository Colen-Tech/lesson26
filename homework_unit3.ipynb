```python
def calculate_bmi(weight_kg, height_m):
    """Calculates BMI given weight in kg and height in meters."""
    if height_m <= 0:
        return None
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 2)

def get_bmi_category(bmi):
    """Returns the standard adult BMI category."""
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 24.9:
        return "Normal weight"
    elif 25 <= bmi < 29.9:
        return "Overweight"
    else:
        return "Obesity"

def main():
    print("Welcome to the BMI Calculator")
    try:
        age = int(input("Enter your age (e.g., 25): "))
        gender = input("Enter your gender (e.g., Male, Female, Other): ").strip().capitalize()
        weight = float(input("Enter your weight in kg (e.g., 65.5): "))
        height = float(input("Enter your height in meters (e.g., 1.75): "))

        bmi_value = calculate_bmi(weight, height)

        if bmi_value is not None:
            category = get_bmi_category(bmi_value)

            print("\n--- Your Results ---")
            print(f"Age: {age}")
            print(f"Gender: {gender}")
            print(f"BMI: {bmi_value}")
            print(f"Category: {category}")

            # Add a disclaimer for users under 18
            if age < 18:
                print("\n*Note: Standard BMI categories are designed for adults (18+).")
                print(" For children and teens, doctors typically use age and gender-specific growth charts.")

        else:
            print("\nError: Height must be greater than zero.")

    except ValueError:
        print("\nInvalid input. Please make sure you are entering numbers for age, weight, and height.")

if __name__ == "__main__":
    main()
```
